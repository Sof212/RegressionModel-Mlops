faison un test stat

```python
import pandas as pd
from scipy.stats import f_oneway

# Rassembler X et y dans un DataFrame
df_test = X[categorical_cols].copy()
df_test["target"] = y

results = []

for col in categorical_cols:
    groups = [
        group["target"].dropna().values
        for _, group in df_test.groupby(col, observed=True)
    ]

    # Il faut au moins 2 groupes non vides
    groups = [g for g in groups if len(g) > 0]

    if len(groups) >= 2:
        f_stat, p_value = f_oneway(*groups)

        results.append({
            "Variable": col,
            "F-statistic": f_stat,
            "p-value": p_value,
            "Nb_categories": len(groups)
        })

# Trier les résultats par p-value croissante
anova_results = pd.DataFrame(results)
anova_results = anova_results.sort_values("p-value")

print(anova_results.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import joblib

from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

# Ordre des catégories ordinales
quality = ["Po", "Fa", "TA", "Gd", "Ex"]

ordinal_cols = [
    "ExterQual",
    "KitchenQual",
    "BsmtQual",
    "GarageFinish",
    "CentralAir"
]

ordinal_categories = [
    quality,
    quality,
    quality,
    ["Unf", "RFn", "Fin"],
    ["N", "Y"]
]

# Colonnes nominales
nominal_cols = ["Foundation", "Neighborhood"]

# 1. Initialiser les encodeurs
ord_enc = OrdinalEncoder(
    categories=ordinal_categories,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

ohe_enc = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# 2. Appliquer les encodages sur data_test
data_test_encoded = data_test.copy()

data_test_encoded[ordinal_cols] = ord_enc.fit_transform(
    data_test[ordinal_cols]
)

nominal_encoded = ohe_enc.fit_transform(
    data_test[nominal_cols]
)

# 3. Créer un DataFrame pour les colonnes One-Hot
nominal_encoded_df = pd.DataFrame(
    nominal_encoded,
    columns=ohe_enc.get_feature_names_out(nominal_cols),
    index=data_test.index
)

# 4. Remplacer les colonnes nominales originales
data_test_encoded = data_test_encoded.drop(columns=nominal_cols)

data_test_encoded = pd.concat(
    [data_test_encoded, nominal_encoded_df],
    axis=1
)

# 5. Enregistrer les encodeurs
joblib.dump(ord_enc, "/Users/sofianeagounikaci/RegressionModel-Mlops/mlartifactsordinal_encoder.pkl")
joblib.dump(ohe_enc, "/Users/sofianeagounikaci/RegressionModel-Mlops/mlartifactsonehot_encoder.pkl")

# 6. Enregistrer le dataset transformé (facultatif)
data_test_encoded.to_csv("../data/processed/data_test_encoded.csv", index=False)

print("Encodage terminé.")
print("Encodeurs enregistrés.")
print(data_test_encoded.head())
```

```python

import numpy as np
import mlflow
import mlflow.lightgbm
import mlflow.sklearn

from lightgbm import LGBMRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)


mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("Machine Learning regression")


def evaluate_model(
    model,
    model_name,
    params,
    X_train,
    y_train,
    X_test,
    y_test
):
    if mlflow.active_run() is not None:
        mlflow.end_run()

    with mlflow.start_run(run_name=model_name):

        mlflow.log_params(params)
        mlflow.log_param("model_type", model_name)
        mlflow.log_param("target_transform", "log1p")


        model.fit(X_train, np.log1p(y_train))

        y_pred = np.expm1(model.predict(X_test))


        rmse = np.sqrt(mean_squared_error(y_test, y_pred))
        mae = mean_absolute_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)


        rmse_log = np.sqrt(
            mean_squared_error(
                np.log1p(y_test),
                np.log1p(y_pred)
            )
        )

        mlflow.log_metrics({
            "rmse": rmse,
            "mae": mae,
            "r2": r2,
            "rmse_log": rmse_log
        })


        if model_name == "LightGBM":
            mlflow.lightgbm.log_model(
                lgb_model=model,
                name="model"
            )
        else:
            mlflow.sklearn.log_model(
                sk_model=model,
                name="model",
                serialization_format="pickle"
            )

        print(f"\n--- {model_name} ---")
        print(f"RMSE     : {rmse:,.2f}")
        print(f"MAE      : {mae:,.2f}")
        print(f"R²       : {r2:.4f}")
        print(f"RMSE log : {rmse_log:.4f}")

        return {
            "model": model_name,
            "rmse": rmse,
            "mae": mae,
            "r2": r2,
            "rmse_log": rmse_log
        }



lgbm_params = {
    "n_estimators": 1500,
    "learning_rate": 0.03,
    "max_depth": -1,
    "num_leaves": 31,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "random_state": 42,
    "n_jobs": -1,
    "verbosity": -1
}

lgbm_model = LGBMRegressor(**lgbm_params)

results_lgbm = evaluate_model(
    model=lgbm_model,
    model_name="LightGBM",
    params=lgbm_params,
    X_train=X_train_enc,
    y_train=y_train,
    X_test=X_test_enc,
    y_test=y_test
)


rf_params = {
    "n_estimators": 500,
    "max_depth": None,
    "min_samples_split": 2,
    "min_samples_leaf": 1,
    "max_features": 1.0,
    "random_state": 42,
    "n_jobs": -1
}

rf_model = RandomForestRegressor(**rf_params)

results_rf = evaluate_model(
    model=rf_model,
    model_name="RandomForest",
    params=rf_params,
    X_train=X_train_enc,
    y_train=y_train,
    X_test=X_test_enc,
    y_test=y_test
)
```

Optimisation des Hyperametres xgboost

```python

import optuna
import numpy as np
import mlflow
import mlflow.xgboost

from xgboost import XGBRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)


mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("Machine Learning regression")


def objective(trial):

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators", 300, 2000
        ),
        "learning_rate": trial.suggest_float(
            "learning_rate", 0.005, 0.1, log=True
        ),
        "max_depth": trial.suggest_int(
            "max_depth", 3, 10
        ),
        "subsample": trial.suggest_float(
            "subsample", 0.6, 1.0
        ),
        "colsample_bytree": trial.suggest_float(
            "colsample_bytree", 0.6, 1.0
        ),
        "min_child_weight": trial.suggest_float(
            "min_child_weight", 1, 10
        ),
        "gamma": trial.suggest_float(
            "gamma", 0, 5
        ),
        "reg_alpha": trial.suggest_float(
            "reg_alpha", 1e-5, 10, log=True
        ),
        "reg_lambda": trial.suggest_float(
            "reg_lambda", 1e-5, 10, log=True
        ),
        "random_state": 42,
        "n_jobs": -1,
        "objective": "reg:squarederror",
        "eval_metric": "rmse"
    }

    model = XGBRegressor(**params)


    model.fit(X_train_enc, np.log1p(y_train))


    pred_log = model.predict(X_test_enc)


    rmse_log = np.sqrt(
        mean_squared_error(np.log1p(y_test), pred_log)
    )

    return rmse_log


study = optuna.create_study(
    direction="minimize",
    study_name="xgboost_ames_housing"
)

study.optimize(objective, n_trials=60)

print("Meilleurs paramètres :", study.best_params)
print("Meilleur RMSE log :", study.best_value)


best_params = study.best_params.copy()
best_params.update({
    "random_state": 42,
    "n_jobs": -1,
    "objective": "reg:squarederror",
    "eval_metric": "rmse"
})

best_model = XGBRegressor(**best_params)

best_model.fit(X_train_enc, np.log1p(y_train))


y_pred = np.expm1(best_model.predict(X_test_enc))


rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

rmse_log = np.sqrt(
    mean_squared_error(
        np.log1p(y_test),
        np.log1p(np.maximum(y_pred, 0))
    )
)

print("\n--- XGBoost optimisé ---")
print(f"RMSE     : {rmse:,.2f}")
print(f"MAE      : {mae:,.2f}")
print(f"R²       : {r2:.4f}")
print(f"RMSE log : {rmse_log:.4f}")



if mlflow.active_run() is not None:
    mlflow.end_run()

with mlflow.start_run(run_name="XGBoost_Optuna_best"):

    mlflow.log_params(best_params)
    mlflow.log_param("optimization", "Optuna")
    mlflow.log_param("n_trials", len(study.trials))
    mlflow.log_param("target_transform", "log1p")

    mlflow.log_metrics({
        "rmse": rmse,
        "mae": mae,
        "r2": r2,
        "rmse_log": rmse_log
    })

    mlflow.xgboost.log_model(
        xgb_model=best_model,
        name="xgb_model"
    )

print("\nModèle optimisé enregistré dans MLflow.")
```
